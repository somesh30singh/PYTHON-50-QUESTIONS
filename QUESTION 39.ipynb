# Question 39: Convert List of Tuples to Dictionary

```python
list1 = [("a", 1), ("b", 2), ("c", 3), ("d", 4)]
dict_word = {}
for key,value in list1:
    dict_word[key]= value
print(dict_word)
```

```python
def convert_dict(list2 : list) -> dict:
    """
        Converts a list of tuples into a dictionary.

        Each tuple in the list should contain exactly two elements:
        the first element is treated as the key and the second as the value.
    
        Parameters:
        tuple_list (list): A list containing tuples of key-value pairs.
    
        Returns:
        dict: A dictionary created from the list of tuples.
    
        Raises:
        TypeError: If the input is not a list.
        TypeError: If any item in the list is not a tuple.
        ValueError: If any tuple does not contain exactly two elements.
    """
    if not isinstance(list2,list):
        raise TypeError("INPUT MUST BE A LIST")
    if len(list2) == 0:
        raise ValueError("LIST SHOULD NOT BE EMPTY")

    final_dict = {}
    for key,value in list2:
        final_dict[key] =  value
    return final_dict

if __name__ =="__main__":
    try:
        user_input = input("PLEASE ENTER LIST TO CONVERT :").title()
        if not user_input:
            raise TypeError("LIST SHOULD NOT BE EMPTY")
        items = user_input.split(",")

        user_input = [tuple(item.strip().split()) for item in items]
        result = convert_dict(user_input)
        print(result)

    except(TypeError,ValueError) as error:
        print(error)
    except Exception as error:
        print(error)
```
