# Question 31: Calculate Sum of Squares of First n Natural Numbers

```python
number = 5000000000
print(number*(number+1)*(2*number+1)//6)
```

```python
def sum_of_square(numb: int) -> int:
    """
            Calculates the sum of squares of the first n natural numbers.
        
        Parameters:
        n (int): The total number of natural numbers.
        
        Returns:
        int: The sum of squares of the first n natural numbers.
             Returns None if an error occurs.
        
        The program uses loop logic to calculate the square of
        each natural number and adds the values to obtain the
        final result.
        
        Exception Handling:
        --> ValueError: Raised when the input value is negative or invalid.
        --> TypeError: Raised when the input is not an integer type.
    """
    if not isinstance(numb,int):
        raise TypeError("INPUT MUST BE AN ENTEGER")
    if numb <1:
        raise ValueError("INPUT SHOULD NOT BE NEGATIVE NUMBER")

    Total = (numb*(numb+1)*(2*numb+1)//6)
    return Total

if __name__=="__main__":
    try:
        user_input = input("PLEASE ENTER YOUR NUMBER TO CHECK :")
        if not user_input.isdigit():
            raise TypeError("ERROR : PLEASE ENTER VALID INPUT")
        user_input = int(user_input)

        result = sum_of_square(user_input)
        print(result)

    except(TypeError,ValueError) as error:
        print(error)
    except Exception as error:
        print(error)
```
