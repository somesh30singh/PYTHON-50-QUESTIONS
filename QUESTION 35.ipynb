```python
def power_of_two(numb: int) -> bool:
    """
        Check whether a given number is a power of 2.

        A number is considered a power of 2 if it can be expressed as
        2 raised to an integer exponent (e.g., 1, 2, 4, 8, 16, 32, ...).
    
        Parameters:
            num (int): The number to be checked.
    
        Returns:
            bool:
                True if the number is a power of 2.
                False otherwise.
    
        Raises:
            TypeError: If the input is not an integer.
            
    """
    if not isinstance(numb,int):
        raise TypeError("INPUT MUST BE AN INTEGER")
    if numb <= 0:
        return False

    if numb>0 and (numb & (numb-1)) == 0:
        return True, (numb).bit_length()-1
    else:
        return False, None

if __name__ == "__main__":
    try:
        user_input = input("PLEASE ENTER YOUR NUMBER :")
        user_input = int(user_input)

        result = power_of_two(user_input)
        print(result)

    except TypeError as error:
        print(error)
    except Exception as error:
        print(error)
```
