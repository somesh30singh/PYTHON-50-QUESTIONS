```python
def find_LCM(num:int, numb: int) -> int:
    """
        Finds the Least Common Multiple (LCM) of two numbers.

        Parameters:
        a (int): The first integer.
        b (int): The second integer.
    
        Returns:
        int: The Least Common Multiple of the two numbers.
    
        Raises:
        TypeError: If either input is not an integer.
        ValueError: If either number is zero.
        
    """
    if not isinstance(num,int) or not isinstance(numb,int):
        raise TypeError("INPUT MUST BE AN INTEGER")

    if num == 0 or numb == 0:
        raise ValueError("INPUT CANNOT DEFINE AS ZERO")

    def GCD(num,numb):
        while numb !=0:
            num,numb = numb,num%numb
        return num

    return(abs(num*numb)//GCD(num,numb))

if __name__=="__main__":
    try:
        number_A = input("PLEASE ENTER NUMBER :")
        number_B = input("PLEASE ENTER NUMBER :")

        number_A = int(number_A)
        number_B = int(number_B)
    except ValueError:
        print("ERROR : PLEASE ENTER VALID INPUT")
    else:
        try:
            result = find_LCM(number_A,number_B)
            print(result)
    
        except(TypeError,ValueError) as error:
            print(error)
        except Exception as error:
            print(error)
```
