# Question 47: Find the First Non-Repeated Character in a String

```python
def unique_alphabet(string1 : str) -> list:
    """
        Finds the first non-repeated character in a string.

        Parameters:
        text (str): The input string to search for a non-repeated character.
    
        Returns:
        str: The first non-repeated character in the string.
        None: If all characters are repeated or the string is empty.
    
        Raises:
        TypeError: If the input is not a string.
        ValueError: if the string value is empty.

    """
    if not isinstance(string1,str):
        raise TypeError("INPUT MUST BE A STRING")
    if len(string1) == 0:
        raise ValueError("INPUT SHOULD NOT BE EMPTY")

    text_dict = {}
    for char in string1:
        if char in text_dict:
            text_dict[char] +=1
        else:
            text_dict[char] =1
    for chara in string1:
        if text_dict[chara] == 1:
            return chara

if __name__ == "__main__":
    try:
        user_input =  input("PLEASE ENTER YOUR STRING TO CHECK:")

        result = unique_alphabet(user_input)
        print(result)

    except(TypeError,ValueError) as error:
        print(error)
    except Exception as error:
        print(error)
```
