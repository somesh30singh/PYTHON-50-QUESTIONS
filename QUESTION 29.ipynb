# Question 29: Count Frequency of Each Word in a Text File

```python
import string
def count_frequency():
    """
        This program reads a text file and counts the frequency
        of each word present in the file. It then displays the
        word count in a readable format.
    """
    
    with open(r"D:\SOMESH FILES\PYTHON CLASS\PYTHON FILES\ASSIGNMENTS AND TEST OR OTHERS\QUESTION 28.txt", mode = "r") as file:
        content = file.read()
    content = content.lower()
    pattern = str.maketrans("","", string.punctuation)
    clean_data =  content.translate(pattern).split()
    
    dict_word_count = {}
    for item in clean_data:
        if item not in dict_word_count:
            dict_word_count[item] =1
        else:
            dict_word_count[item] = dict_word_count[item] +1

    for key,value in dict_word_count.items():
        print(key,value)

if __name__ == "__main__":
    try:
        print(count_frequency())

    except Exception as error:
        print(error)
```
