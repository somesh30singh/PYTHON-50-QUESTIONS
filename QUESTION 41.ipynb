```python
def sorted_dictionary(dict2 : dict) -> dict:
    """
        Sorts a dictionary by its values.

        Parameters:
        data (dict): The dictionary to be sorted.
        reverse (bool, optional): If True, sorts in descending order.
                                  Default is False (ascending order).
    
        Returns:
        dict: A new dictionary sorted by value.
    
        Raises:
        TypeError: If the input is not a dictionary.
        ValueError: If the dictionary is empty
    """
    if not isinstance(dict2,dict):
        raise TypeError("INPUT MUST BE A DICTIONARY")
    if len(dict2) == 0:
        raise ValueError("DICTIONARY SHOULD NOT BE EMPTY")

    items = list(dict2.items())
    for i in range(len(items)):
        for j in range(0,len(items)-1-i):
            if items[j][1] > items[j+1][1]:
                items[j],items[j+1] = items[j+1],items[j]
    return items

if __name__ =="__main__":
    try:
        user_input = {'apple': 5, 'banana': 2, 'cherry': 8, 'date': 3}

        result =  sorted_dictionary(user_input)
        print(result)

    except(TypeError,ValueError) as error:
        print(error)
    except Exception as error:
        print(error)
```
