```python
def check_is_armstrong(number1 : int) -> bool:
    """
        Checks whether a number is an Armstrong number.

        An Armstrong number is a number that is equal to the sum
        of its own digits each raised to the power of the number
        of digits.
    
        Parameters:
        num (int): The number to check.
    
        Returns:
        bool: True if the number is an Armstrong number, otherwise False.
    
        Raises:
        TypeError: If the input is not an integer.
        ValueError: If the number is negative.

    """
    if not isinstance(number1,int):
        raise TypeError("INPUT MUST BE AN INTEGER")
    if number1 == 0:
        raise ValueError("INPUT SHOULD NOT BE ZERO")

    temp_empt = number1
    num_count = 0
    while temp_empt > 0:
        num_count = num_count + 1
        temp_empt = temp_empt//10

    temp_empt = number1
    final_result = 0
    while temp_empt > 0:
        digit = temp_empt % 10
        final_result = final_result + digit ** num_count
        temp_empt = temp_empt // 10
    return final_result == number1

if __name__ == "__main__":
    try:
        user_input =  input("PLEASE ENTER YOUR NUMBER TO CHECK :")
        if not user_input.isdigit():
            raise TypeError("ERROR : PLEASE ENTER VALID INPUT")
        user_input =  int(user_input)
        result = check_is_armstrong(user_input)
        if result == True:
            print(f"{result} : THIS NUMBER IS ARMSTRONG")
        else:
            print(f"{result} : THIS NUMBER IS NOT ARMSTRONG")
    except(TypeError,ValueError) as error:
        print(error)
    except Exception as error:
        print(error)
```
