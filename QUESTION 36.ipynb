# Question 36: Flatten a Nested List

```python
nested_list = [1, [2, [3, 4], 5], 6]
result = []
empty_list = list(nested_list)

while empty_list:
    items = empty_list.pop()
    if isinstance(items,list):
        empty_list.extend(items)
    else:
        result.append(items)
print(result[::-1])
```

```python
def nested_lists(list1 : list) -> list:
    """
        Flattens a nested list into a single-dimensional list.
    
        This function recursively traverses all elements in the input list.
        If an element is itself a list, its contents are flattened and added
        to the final result.
    
        Parameters:
        nested_list (list): A list that may contain nested lists and elements
                            of any data type.
    
        Returns:
        list: A flattened list containing all elements from the nested structure.
    
        Raises:
        TypeError: If the input is not a list.
        ValueError: If the list is empty raise the error.
        """
    if not isinstance(list1,list):
        raise TypeError("INPUT MUST BE A LIST")
    if len(list1) == 0:
        raise ValueError("LIST SHOULD NOT BE EMPTY")

    empty_lists = list(list1)
    final_result = []

    while empty_lists:
        items = empty_lists.pop()
        if isinstance(items,list):
            empty_lists.extend(items)
        else:
            final_result.append(items)
    return final_result[::-1]

if __name__ == "__main__":
    try:
        list1 = [1, [2, [3, 4], 5], 6]

        result = nested_lists(list1)
        print(result)
    except(TypeError,ValueError) as error:
        print(error)
    except Exception as error:
        print(error)
```
