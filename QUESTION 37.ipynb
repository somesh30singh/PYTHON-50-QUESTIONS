# Question 35: Find Pairs with Given Sum in a List

```python
List = [1, 2, 3, 4, 5, 6,8]
target = 7
result = []

for i in range(0,len(List)):
    for j in range(i+1,len(List)):
        if List[i] + List[j] == target:
            result.append((List[i],List[j]))
print(result)
```

```python
def find_pairs(list1: list, target : int) -> list:
    """
        Finds all pairs of numbers in a list whose sum equals the target value.
    
        This function checks every possible pair in the list and returns
        the pairs that add up to the specified target sum.
    
        Parameters:
        numbers (list): A list of integers or floating-point numbers.
        target_sum (int | float): The target sum for which pairs are searched.
    
        Returns:
        list: A list of tuples, where each tuple contains a valid pair
              whose sum equals the target value.
    
        Raises:
        TypeError: If `numbers` is not a list.
        TypeError: If `target_sum` is not a numeric value.
        ValueError: If the list contains non-numeric elements.
    """

    if not isinstance(list1,list):
        raise TypeError("INPUT MUST BE A LIST")
    if not isinstance(target,int):
        raise TypeError("INPUT MUST BE AN INTEGER")
    if not all(isinstance(item,(int)) for item in list1):
        raise TypeError("LIST MUST CONTAINS INTEGER VALUE ONLY")
    if len(list1) == 0:
        raise ValueError("LIST SHOULD NOT BE EMPTY")
    
    final_result = []
    for i in range(len(list1)):
        for j in range(i+1,len(list1)):
            if list1[i] + list1[j] == target:
                final_result.append((list1[i],list1[j]))
    return final_result

if __name__ == "__main__":
    try:
        user_input =  input("PLEASE ENTER YOUR LIST :")
        user_input = [int(item) for item in user_input.split()]
        
        target = input("ENTER TARGET :")
        target = int(target)
    except ValueError:
        print("ERROR : PLEASE ENTER VALID INPUT")
    else:
        try:
            result = find_pairs(user_input,target)
            print(result)
    
        except(TypeError,ValueError) as error:
            print(error)
        except Exception as error:
            print(error)
```
