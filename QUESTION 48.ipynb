# Question 48: Check if String Contains Only Digits.

```python
string = "111a"

for char in string:
    if char < "0" or char > "9":
        print(False,": STRING DOES NOT CONTAINS INETEGER VALUE ONLY")
        break
else:
    print(True, ": STRING CONTAINS INTEGER VALUE ONLY")
```

```python
def check_int_only(string1 : str) -> bool:
    """
        Finds the first non-repeated character in a string.

        Parameters:
        text (str): The input string to search for a non-repeated character.
    
        Returns:
        str: The first non-repeated character in the string.
        None: If all characters are repeated or the string is empty.
    
        Raises:
        TypeError: If the input is not a string.
        ValueError: if the string is empty.

    """
    if not isinstance(string1,str):
        raise TypeError("INPUT MUST BE A STRING")
    if len(string1) == 0:
        raise ValueError("INPUT SHOULD NOT BE EMPTY")

    for chara in string1:
        if chara < "0" or chara > "9":
            return False

    return True

if __name__ == "__main__":
    try:
        string_user = input("PLEASE ENTER STRING TO CHECK :")

        result = check_int_only(string_user)
        if result:
            print(result," : THIS STRING CONTAINS INTEGERS ONLY")
        else:
            print(result,": THIS STRING NOT CONTAINS INTEGERS ONLY")

    except (TypeError,ValueError) as error:
        print(error)
    except Exception as error:
        print(error)
```
