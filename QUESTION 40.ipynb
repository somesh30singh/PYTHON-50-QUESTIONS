# Question 40: Find the Median of a List of Numbers.

```python
list1 = [1, 3, 5, 9, 7,8]
list1.sort()
print(list1)
length = len(list1)
print(length)
mid = length //2
print(mid)

if length %2 != 0:
    print(list1[mid])
else:
    print((list1[mid-1]+list1[mid])/2)
```

```python
def find_median(list2 : list) -> int:
    """
        Finds the median value of a list of numbers.

        The median is the middle value in a sorted list. If the list
        contains an even number of elements, the median is calculated
        as the average of the two middle values.
    
        Parameters:
        numbers (list): A list of integers or floating-point numbers.
    
        Returns:
        int | float: The median value of the list.
    
        Raises:
        TypeError: If `numbers` is not a list.
        ValueError: If the list is empty.
        ValueError: If the list contains non-numeric elements.

    """
    if not isinstance(list2,list):
        raise TypeError("INPUT MUST BE A LIST")
    if not all(isinstance (item,(int)) for item in list2):
        raise TypeError("LIST MUST CONATINS INTEGER VALUE ONLY")
    if len(list2) == 0:
        raise ValueError("LIST SHOULD NOT BE EMPTY")

    list2.sort()
    length_list = len(list2)
    mid = length_list//2

    if length_list % 2 != 0:
        return list2[mid]
    else:
        return (list2[mid-1] + list2[mid])/2

if __name__== "__main__":
    try:
        user_input = input("PLEASE ENTER YOUR LIST :")

        user_input = [int(item) for item in user_input.split()]
    except ValueError:
        print("ERROR : PLEASE ENTER VALID INPUT")
    else:
        try:
            result = find_median(user_input)
            print(result)
        except(TypeError,ValueError) as error:
            print(error)
        except Exception as error:
            print(error)
```
