# Question 45: Find Common Elements in Three Lists

```python
list1 = [1, 2, 3, 4, 5, 5, 6]
list2 = [3, 4, 5, 6, 7]
list3 = [5, 6, 7, 8, 9]

common_element = []

for item in list1:
    if item in list2 and item in list3 and item not in common_element:
        common_element.append(item)
print(common_element)
```

```python
def common_elements(list1:list,list2:list,list3:list) -> list:
    """
        Finds the common elements present in three lists.

        Parameters:
        list1 (list): The first list of elements.
        list2 (list): The second list of elements.
        list3 (list): The third list of elements.
    
        Returns:
        list: A list containing the common elements found in all three lists.
    
        Raises:
        TypeError: If any of the inputs is not a list.
        ValueError: If any of the lists is empty.
        
    """
    if not isinstance(list1,list) or not isinstance(list2,list) or not isinstance(list3,list):
        raise TypeError("INPUT MUST BE A LIST")
    if len(list1) == 0 or len(list2) == 0 or len(list3) == 0:
        raise ValueError("NOT A SINGLE LIST SHOULD BE EMPTY")

    common_object = []

    for item in list1:
        if item in list2 and item in list3 and item not in common_object:
            common_object.append(item)
    return common_object

if __name__=="__main__":
    try:
        list1_user = input("PLEASE ENTER :").lower().split()
        list2_user = input("PLEASE ENTER :").lower().split()
        list3_user = input("PLEASE ENTER :").lower().split()

        result = common_elements(list1_user,list2_user,list3_user)
        print(result)

    except(TypeError, ValueError) as error:
        print(error)
    except Exception as error:
        print(error)
        
```
