```python
number = 123456
result = 0
sign = -1 if number < 0 else 1
number = abs(number)

while number > 0:
    digit = number % 10
    result = result * 10+digit
    number = number // 10
print(result)
```

```python
def reverse_integer(numb : int) -> int:
    """
            Reverses the digits of a given integer and returns the reversed number.
    
        Parameters:
        num (int): The integer whose digits need to be reversed.
        
        Returns:
        int: The integer with digits reversed.
             Returns None if an error occurs.
        
        The function uses loop logic to extract digits one by one
        and reverse their order to form the final number.
        
        Exception Handling:
        TypeError: Raised when the input is not an integer type.
    """
    if not isinstance(numb,int):
        raise TypeError("INPUT MUST BE AN INTEGER")
        
    reverse_int = 0
    sign = -1 if numb<0 else 1
    numb = abs(numb)
    while numb>0:
        digit = numb% 10
        reverse_int = reverse_int * 10 +digit
        numb = numb // 10
    return reverse_int

if __name__=="__main__":
    try:
        user_input = input("PLEASE ENTER NUMBER TO REVERSE :")
        user_input = int(user_input)

    except ValueError:
        print("ERROR : PLEASE ENTER VALID INPUT")

    else:
        try:
            result = reverse_integer(user_input)
            print(result)

        except(TypeError,ValueError) as error:
            print(error)
        except Exception as error:
            print(error)
```
