# Question 30:  Find Missing Number in Consecutive List

```python
List = [1,2,3,4,5,6,15] # missing numbers [7,8,9,10,11,12,13,14]
missing_value = []

for i in range(0,len(List)-1):
    sample = List[i]
    next_num = List[i+1]

    if next_num - sample != 1:
        for item in range(sample +1,next_num):
            missing_value.append(item)
print(missing_value)
```

```python
def find_missing_value(list1 : list) -> list:
    """
        Finds the missing number in a consecutive list of numbers.
        
        Parameters:
        numbers (list): A list containing consecutive numbers with one missing value.
        
        Returns:
        int: The missing number from the sequence.
             Returns None if no missing number is found.

        Exception Handling:
         --> TypeError: Input must be a list
         --> ValueError: List should not be empty
        
    """
    if not isinstance(list1,list):
        raise TypeError("INPUT MUST BE A LIST")
    if len(list1) == 0:
        raise ValueError("LIST SHOULD NOT BE EMPTY")
    
    missing_number = []
    for i in range(0,len(list1)-1):
        simple = list1[i]
        next_num = list1[i+1]

        if next_num - simple != 1:
            for item in range(simple+1,next_num):
                missing_number.append(item)
    return missing_number

if __name__ == "__main__":
    try:
        list1_input = input("PLEASE ENTER YOUR LIST TO FIND MISSING NUMBER :")

        list1_input = [int(item) for item in list1_input.split()]
    except ValueError:
        print("ERROR : PLEASE ENTER VALID INPUT")
    else:
        try:
            result = find_missing_value(list1_input)
            print(result)
    
        except (TypeError,ValueError) as error:
            print(error)
        except Exception as error:
            print(error)
```
