```python
import string
def find_longest_word(sentence : str) -> str:
    """
        Finds the longest word present in a given sentence.

        Parameters:
        sentence (str): The sentence in which the longest word
        needs to be searched.
        
        Returns:
        str: The longest word in the sentence.
             Returns None if an error occurs.
        
        The function uses loop logic to compare the length of
        each word in the sentence and identify the longest word.
        
        Exception Handling:
        ValueError: Raised when the sentence is empty or invalid.
        TypeError: Raised when the input is not a string type.
    """
    if not isinstance(sentence,str):
        raise TypeError("INPUT MUST BE A STRING")
    if len(sentence) ==0:
        raise ValueError("INPUT SHOULD NOT BE EMPTY")

    sentence = sentence.lower()
    pattern = str.maketrans("","",string.punctuation)
    clean_text = sentence.translate(pattern).split()
    
    word_dict_count = {}
    for item in clean_text:
        if item not in word_dict_count:
            word_dict_count[item] = len(item)
    
    max_value = max(word_dict_count.values())
    
    longest_word = [key for key,value in word_dict_count.items() if value == max_value]
    return longest_word

if __name__ =="__main__":
    try:
        user_input = input("PLEASE ENTER YOUR SENTENCE :")
        if user_input.isdigit():
            raise TypeError("ERROR : PLEASE ENTER VALID INPUT")
        result = find_longest_word(user_input)

        print(result)
    except(TypeError,ValueError) as error:
        print(error)
```
