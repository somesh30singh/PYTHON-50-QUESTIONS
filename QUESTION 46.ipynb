# Question 46: Remove Nth Occurrence of a Character from a String

```python
string = "this is python class"
remove_character = "s"
int_value = 2

occurance_count = 0
result = []
for char in string:
    if char == remove_character:
        occurance_count = occurance_count +1
        if occurance_count == int_value:
            continue
    result.append(char)
print("".join(result))
```

```python
def remove_nth_occurance(string:str,char_to_remove:str,n:int) -> str:
    """
        Removes the nth occurrence of a character from a string.

        Parameters:
        text (str): The input string.
        char (str): The character to remove.
        n (int): The occurrence number to remove.
    
        Returns:
        str: A new string with the nth occurrence of the character removed.
        str: Original string if the nth occurrence does not exist.
    
        Raises:
        TypeError: If text or char is not a string.
        TypeError: If n is not an integer.
        ValueError: If char is not a single character.
        ValueError: If n is less than or equal to 0.
        
    """
    if not isinstance(string,str) or not isinstance(char_to_remove,str):
        raise TypeError("INPUT MUST BE A STRING")
    if len(string) == 0 or len(char_to_remove) == 0:
        raise ValueError("STRING AND CHARACTER INPUT SHOULD NOT BE EMPTY")


    occurence_count = 0
    result = []
    for current_count in string:
        if current_count == char_to_remove:
            occurence_count = occurence_count +1
            if occurence_count == n:
                continue
        result.append(current_count)
    return "".join(result)

if __name__ == "__main__":
    try:
        user_input = input("PLEASE ENTER YOUR SENTENCE :")
        user_input1 = input("PLEASE ENTER CHARACTER TO REMOVE FROM THE SENTENCE :")
        int_input = int(input("ENTER CHARACTER'S POSITION VALUE :"))

        result = remove_nth_occurance(user_input,user_input1,int_input)
        print(result)
    except(TypeError,ValueError) as error:
        print(error)
    except Exception as error:
        print(error)
```
