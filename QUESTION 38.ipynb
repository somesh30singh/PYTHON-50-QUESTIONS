# Question 38: Implement Queue using List.

```python
class Queue:
    def __init__(self):
        self.queue = []

    def is_empty(self):
        return len(self.queue) == 0

    def enqueue(self,value):
        self.queue.append(value)
        print(f"{value} ENQUEUED. QUEUED {self.queue}")

    def dequeue(self):
        if self.is_empty():
            print("QUEUE IS EMPTY, CANNOT DEQUEUE")
        else:
            removed = self.queue.pop(0)
            print(f"{removed} DEQUEUED. QUEUED {self.queue}")
    def peek(self):
        if self.is_empty():
            print("QUEUE IS EMPTY")
        else:
            print(f"FRONT ELEMENT : {self.queue[0]}")

q = Queue()
```
