# Question 33: Implement a Stack using List

```python
class Stack:
    def __init__(self):
        self.stack = []

    def push(self,value):
        self.stack.append(value)
        print(value)
    def pop(self):
        if self.is_empty():
            return None
        return self.stack.pop()
    def peek(self):
        if self.is_empty():
            return None
        return self.stack[-1]
    def is_empty(self):
        return len(self.stack) == 0
    def __str__(self):
        return str(self.stack)

if __name__ == "__main__":
    stack = Stack()
```

```python
stack.push(10)
```

```python
stack.peek()
```

```python
print(stack)
```
